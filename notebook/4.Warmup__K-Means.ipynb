````python
import pandas as pd

sheet_id = '1a06tAjq1ZRYI2rkMj5Tl2YKN2KpXkZxiI9we72LfmJU'
sheet_gid = "0"
url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=csv&gid={sheet_gid}"
foods_df = pd.read_csv(url, index_col="student").drop(columns=['age', 'favorite_color'])
foods_df
````

####Exercise 1: Import the K-Means model and initialize it with a random state and number of clusters

````python
from sklearn.cluster import KMeans

k = 3  # number of clusters
kmeans = KMeans(n_clusters=k, random_state=42)
````

#####Hint:

Take a look at the documentation [here](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.KMeans.html)<br>
At the very top it says `class sklearn.cluster.KMeans(`... This is kind of like a file path telling us where to find the model in the library. So to import it we run `from sklearn.cluster import KMeans`.<br><br>
To initialize the model we call it like a function with parameters and assign the result to a variable: `model = KMeans(param = _____, param = _____)` You can look at the documentation to find what the names are for the parameters that we have to use to initialize it with a random state and number of clusters.

####Exercise 2: Fit the model to the food data imported above

````python
kmeans.fit(foods_df)

print(kmeans.cluster_centers_)
````

#####Hint:

Fit the model to the data using the `.fit()` method

####Exercise 3: Extract the cluster assignments from the model and add them as a new column to `foods_df`

````python
foods_df["cluster"] = kmeans.labels_

foods_df
````

````python
foods_df.groupby("cluster").mean()
````

####Exercise 4: Sort the data frame by the cluster assignments to more easily see the groupings

````python
foods_df = foods_df.sort_values("cluster")
foods_df
````

````python
from sklearn.cluster import KMeans

for k in range(2,6):
    kmeans = KMeans(n_clusters=k, random_state=42
    )
    kmeans.fit(foods_df.drop(columns="cluster"))
    foods_df[f"cluster_{k}"] = kmeans.labels_

foods_df
````
